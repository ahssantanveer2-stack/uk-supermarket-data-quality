# Data Profiling

## Project Overview

### Business Scenario
A retail analytics team has received web-scraped pricing data from five major UK supermarkets. Before the data can be used for reporting, price comparisons, or business analysis, its quality must be assessed to identify any issues that could affect the reliability of the analysis.

### Objective
The objective of this phase is to understand the structure and quality of the datasets before performing any cleaning. This ensures that any changes made during the cleaning process are based on evidence rather than assumptions.

---

## What We Know

Based on the available information, we know that:

- The data has been collected through web scraping.
- There are five separate datasets, one for each supermarket.
- The data contains daily supermarket prices.
- Each dataset includes product names, prices, units, categories, dates, and an own-brand indicator.
- Web-scraped data may contain inconsistencies or quality issues that need to be investigated.

---

## What We Need to Investigate

Before cleaning the data, the following questions need to be answered.

### Completeness
- Are there any missing values?
- Are there empty strings?
- Are important fields incomplete?

### Validity
- Are prices stored as numeric values?
- Are dates in a valid format?
- Are unit values valid?
- Are there any impossible prices?

### Consistency
- Are category names consistent?
- Are units written consistently?
- Are boolean values stored consistently?
- Are there differences in spelling or capitalisation?

### Uniqueness
- Are there duplicate rows?
- Are duplicate products expected?

### Timeliness
- Do all records contain scrape dates?
- Are there unexpected gaps in the dates?

---

## Investigation Plan

The data profiling stage will answer each of these questions before any cleaning is performed. The findings will guide the data cleaning and validation stages of the project.
